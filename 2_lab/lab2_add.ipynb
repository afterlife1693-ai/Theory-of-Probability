### Тема 1. Моделирование геометрической вероятности
> Смоделируйте вычисление геометрической вероятности (на примере из лекции про метание дротиков в игровой круг, площадь которого в два раза меньше площади стены). Показать теоретический расчет и графическую иллюстрацию для рассматриваемого примера.


```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, Polygon
import numpy as np
```

```python
def points(n):
    a = 12
    # нижнее основание [x0, a - x0] на y = 0, верхнее основание [x1, a - x1] на y = a
    x0 = 2   # 1 отступ снизу от края
    x1 = 4   # 2 отступ снизу от края
    fig, ax = plt.subplots(figsize=(5, 5))
    ax = plt.gca()

    trapezia = Polygon([[x0, 0], [a - x0, 0], [a - x1, a], [x1, a]], facecolor='b')
    ax.add_patch(Rectangle((0, 0), 12, 12, facecolor='g'))
    ax.add_patch(trapezia)
    X = np.random.uniform(0, a, n)
    Y = np.random.uniform(0, a, n)
    plt.scatter(X, Y, marker='.', c="black")
    inside = []
    k = (x1 - x0) / a  # наклон боковой стороны
    for x, y in zip(X, Y):
        x_left = x0 + k * y
        x_right = a - x0 - k * y
        if x_left <= x <= x_right:
            inside.append([x, y])
    prob = len(inside) / n

    plt.show()
    print(f"в синюю трапецию из {n} точек попало {len(inside)}. Доля попавших = {prob}")
```

**Теоретический расчёт.**

$$S_{трап} = \frac{b_1 + b_2}{2} \cdot h = \frac{8 + 4}{2} \cdot 12 = 72$$

$$S_{квадрат} = a^2 = 144$$

$$P = \frac{S_{трап}}{S_{квадрат}} = \frac{72}{144} = 0.5$$


```python
points(1000)
```
