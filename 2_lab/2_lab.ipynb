Наудачу взятый телефонный номер состоит из 5 цифр.
Какова вероятность того, что в нем все цифры: а) различные;
б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.

$p(A) = \frac{n}{m} = \frac{9*A^4_9}{9*10^4} = \frac{9!}{4!*10^5}$

```python
np.round(9 * sp.perm(9, 4) / (9*10**4), 5)
```

${P(B)} = \frac{n}{m} = \frac{9}{9*10^4} = \frac{1}{10^4} $

```python
np.round(9 / (9*10**4), 5) # 9 так как цифр десять всего - 1 по усл (0 не может быть 1й цифрой)
```

$P(C) = \frac{n}{m} =  \frac{\bar{A}^5_5}{9*10^4} = \frac{5^5}{9*10^4}$

```python
np.round(5**5 / (9*10**4), 5)
```

В старинной игре в кости необходимо было для выигрыша
получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения l l очков; б) выигрыша.

```python
from itertools import product
```

```python
c = 0
all_events = 6**3
for i in product('123456', repeat = 3):
    if sum([int(char) for char in i]) == 11:
        c += 1
c / all_events
```

```python
c = 0
for i in product('123456', repeat = 3):
    if sum([int(char) for char in i]) > 10:
        c += 1
c / all_events
```
