**1.10**

Буквы Т, Е, И, Я, Р, О написаны на отдельных
карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: 
а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

$P(A) = \frac{1}{A_6^3}$

```python
#A
# P = 1/A(3, 6) 
round(1/sp.perm(6, 3), 5)
```

$P(B) = \frac{1}{6!}$

```python
#B
# P = 1/6!
round(1 / sp.factorial(6), 5)
```

**1.11**


Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных
карточках написаны три буквы А, две буквы Н и одна буква С.

$P(A) = \frac{P_3 \cdot P_2}{P_6} = \frac{3! \cdot 2!}{6!}$

```python
# P = Pблаг / 6!
# Pблаг = 3!*2! (ААА и НН)
round(sp.factorial(3)*sp.factorial(2)/sp.factorial(6), 5)
```

**1.16**

За круглым столом рассаживаются 5 мужчин и
5 женщин. Найти вероятность того, что: а) никакие два лица одного
пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар.

$P(A) = \frac{2(5!)^2}{10!}$

```python
#А
# 5 мужчин сидят через одного, 5 женщин сидят через одного на 1 3 5 7 9 местах, вариантов рассадить мужчин 5!, женщин тоже 5!. 
# тк мужчины могут сидеть и на четных местах, вариантов в 2 раза больше чем 5!**2
# количество всех вариантов = 10! 
round(2*(sp.factorial(5)**2) / sp.factorial(10), 5)
```

$P(B) = \frac{(5!)2^5}{10!}$

```python
#B
round(sp.factorial(5)*2**5/sp.factorial(10), 5)
```

**1.17**

В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

$P(A) = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$

```python
#A
round(sp.comb(9, 2) * sp.comb(8, 7) / sp.comb(36, 7), 5)
```

$P(B) = \frac{C_9^7 \cdot 4^7}{C_{36}^7}$

```python
round(sp.comb(9, 7) * (4 ** 7) / sp.comb(36, 7), 5)
```

```python
round(sp.comb(9, 3) * (6 * sp.comb(4, 4) * sp.comb(4, 2) * sp.comb(4, 1) + 3 * sp.comb(4, 3) * sp.comb(4, 3) * sp.comb(4, 1) + 3 * sp.comb(4, 3) * sp.comb(4, 2) * sp.comb(4, 2)) / sp.comb(36, 7), 5)
```
