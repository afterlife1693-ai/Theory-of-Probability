# Тема 2. Моделирование связи частотного и классического определения вероятности
Приведите пример, показывающий связь частотного и классического определения вероятности с помощью моделирования (на примере задачи про красно-синие шары или задачи о сумме очков на двух кубиках). Графически показать стремление частотной вероятности к классической при увеличении числа опытов.


```python
import numpy as np
from random import sample
import plotly.graph_objs as go
```

```python
def random_comb():
    return sample((['C']*6 + ['D']*6), 6)
comb = random_comb()
comb
```

```python
n_vals = np.array(range(10, 10001, 50))
p_a_vals = []
for n in n_vals:
    n_a = 0
    for _ in range(n):
        comb = random_comb()
        if comb.count('C') == 5:
            n_a += 1
    p_a_vals.append(n_a/n)
            
p = [0.0065*6]*len(n_vals)
fig = go.Figure(layout=dict(height=500, width=800, yaxis=dict(range=[0.01, 0.06])))
fig.add_trace(go.Scatter(x=n_vals, y=p, name='Probability', line_dash='dash'))
fig.add_trace(go.Scatter(x=n_vals, y=p_a_vals, name='Frequency'))
fig.show();
```
