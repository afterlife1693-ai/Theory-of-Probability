```python
import scipy.special as sp
import numpy as np
```

$P(A) = \frac{n}{m} = \frac{1}{5!}$

```python
np.round(1 / sp.factorial(5), 5)
```

$p(A) = \frac{n}{m} = \frac{9*A^4_9}{9*10^4} = \frac{9!}{4!*10^5}$

```python
np.round(9 * sp.perm(9, 4) / (9*10**4), 5)
```

${P(B)} = \frac{n}{m} = \frac{9}{9*10^4} = \frac{1}{10^4} $

```python
np.round(9 / (9*10**4), 5) # 9 так как цифр десять всего - 1 по усл (0 не может быть 1й цифрой)
```

$P(C) = \frac{n}{m} =  \frac{\bar{A}^5_5}{9*10^4} = \frac{5^5}{9*10^4}$

```python
np.round(5**5 / (9*10**4), 5)
```

```python
from itertools import product
```

```python
c = 0
all_events = 6**3
for i in product('123456', repeat = 3):
    if sum([int(char) for char in i]) == 11:
        c += 1
c / all_events
```
