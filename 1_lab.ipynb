**1.16**

$P(A) = \frac{2(5!)^2}{10!}$

```python
#А
# 5 мужчин сидят через одного, 5 женщин сидят через одного на 1 3 5 7 9 местах, вариантов рассадить мужчин 5!, женщин тоже 5!. 
# тк мужчины могут сидеть и на четных местах, вариантов в 2 раза больше чем 5!**2
# количество всех вариантов = 10! 
(2*(sp.factorial(5)**2) / sp.factorial(10))
```
